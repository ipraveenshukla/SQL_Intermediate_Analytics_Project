```python
# Running some operations on 'sales' table
# getting the net revenue by some multiplication
%%sql

SELECT orderdate,
  quantity*netprice*exchangerate AS net_revenue
FROM sales
ORDER BY orderdate desc
limit 10
```

```python
# looking for data available after 1 jan 2024

%%sql

SELECT orderdate,
  quantity*netprice*exchangerate AS net_revenue
FROM sales
WHERE orderdate >= '2024-01-01'
ORDER BY orderdate
```

```python
%%sql
SELECT *
FROM customer
limit 5
```

```python
# joining the tables
%%sql

SELECT s.orderdate,
  s.quantity*s.netprice*s.exchangerate AS net_revenue,
  c.givenname,
  c.surname,
  c.countryfull,
  c.continent
FROM sales s
LEFT JOIN customer c ON s.customerkey = c.customerkey
WHERE orderdate >= '2024-01-01'
ORDER BY s.orderdate desc
limit 10
```

```python
%%sql
SELECT *
FROM product
limit 5
```

```python
# moving forward we want to join one more table to get the product information
%%sql

SELECT s.orderdate,
  s.quantity*s.netprice*s.exchangerate AS net_revenue,
  c.givenname,
  c.surname,
  c.countryfull,
  c.continent,
  p.productname,
  p.categoryname,
  p.subcategoryname
FROM sales s
LEFT JOIN customer c ON s.customerkey = c.customerkey
LEFT JOIN product p ON s.productkey = p.productkey
WHERE orderdate >= '2024-01-01'
ORDER BY s.orderdate desc
limit 10
```

```python
# adding a condition to include 'High' and 'Low' for net_revenue
%%sql

SELECT s.orderdate,
  s.quantity*s.netprice*s.exchangerate AS net_revenue,
  c.givenname,
  c.surname,
  c.countryfull,
  c.continent,
  p.productname,
  p.categoryname,
  p.subcategoryname,
CASE WHEN   s.quantity*s.netprice*s.exchangerate > 1000 THEN 'HIGH' ELSE 'LOW' END AS status
FROM sales s
LEFT JOIN customer c ON s.customerkey = c.customerkey
LEFT JOIN product p ON s.productkey = p.productkey
WHERE orderdate >= '2024-01-01'
ORDER BY s.orderdate desc
limit 10
```
